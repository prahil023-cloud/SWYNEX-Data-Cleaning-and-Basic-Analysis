```python
import pandas as pd

df = pd.read_csv(
    "GSE2034_series_matrix.txt.gz",
    sep="\t",
    compression="gzip",
    comment="!",
    index_col=0
)

print(df.shape)
```

```python
import matplotlib.pyplot as plt

sample_means = df.mean(axis=0)

plt.figure(figsize=(8,5))
plt.hist(sample_means, bins=20)
plt.xlabel("Mean Expression")
plt.ylabel("Number of Samples")
plt.title("GSE2034 - Distribution of Mean Expression")
plt.show()
```

```python
import matplotlib.pyplot as plt

sample_std = df.std(axis=0)

plt.figure(figsize=(8,5))
plt.hist(sample_std, bins=20)
plt.xlabel("Standard Deviation of Expression")
plt.ylabel("Number of Samples")
plt.title("GSE2034 - Distribution of Expression Variability")
plt.show()
```

Interpretation: This histogram shows the variability of gene expression across the 286 samples using standard deviation. It helps identify how much expression values vary between samples. It does not by itself prove a biological difference between patient groups or disease conditions.

```python
sample_means = df.mean(axis=0)
sample_medians = df.median(axis=0)

plt.figure(figsize=(8,5))
plt.scatter(sample_means, sample_medians, alpha=0.6)
plt.xlabel("Mean Expression")
plt.ylabel("Median Expression")
plt.title("GSE2034 - Mean vs Median Expression")
plt.show()
```

Interpretation: This scatter plot compares the mean and median expression values for each sample. Points closer to a consistent relationship indicate that the two summary measures are similar across samples. The plot helps visualize the relationship between mean and median expression, but it does not establish causation or prove differences between biological groups.
